# Model 8 inference
Place `model8_three_seed.joblib` beside this notebook, or attach it as a Kaggle input. Run both cells to score the test sightings and write `submission.csv`.

The competition data folder must contain the extracted metadata and MIEW-ID features, plus the supplied train and test SIFT descriptors. Model 8 uses the labelled train sightings as its identification gallery during inference. The sample submission sets the output row order.

```python
from pathlib import Path

kaggle = Path("/kaggle/input")
repo = next((p for p in (Path.cwd(), *Path.cwd().parents) if (p / "Models/Emad").is_dir()), None)
local = repo / "Models/Emad" if repo else Path.cwd()

# Change these paths if your files are elsewhere.
DATA_DIR = kaggle if kaggle.exists() else (repo or Path.cwd()) / "DATA"
WEIGHTS = next(kaggle.rglob("model8_three_seed.joblib"), None) if kaggle.exists() else local / "model8_three_seed.joblib"
OUTPUT = (Path("/kaggle/working") if kaggle.exists() else Path.cwd()) / "submission.csv"
```

```python
import time
from tempfile import TemporaryDirectory
import joblib

started = time.perf_counter()
if WEIGHTS is None or not Path(WEIGHTS).is_file():
    raise FileNotFoundError("Set WEIGHTS to the supplied model8_three_seed.joblib file.")
package = joblib.load(WEIGHTS)
with TemporaryDirectory() as folder:
    work = Path(folder)
    runtime = {"DATA_DIR": Path(DATA_DIR), "WORK": work}
    exec(package["feature_code"], runtime)
    (work / "public_sift_pairs.npz").write_bytes(package["public_geometry"])
    submission = runtime["score_artifact"](work, package["model"])
submission.to_csv(OUTPUT, index=False)
print(f"Wrote {len(submission):,} rows to {OUTPUT} in {time.perf_counter() - started:.1f} seconds.")
submission.head()
```
